## 2. Check Local Ollama
We need to make sure Ollama is running on your PC (`localhost:11434`).

```python
import requests
import subprocess

try:
    response = requests.get("http://localhost:11434")
    if response.status_code == 200:
        print("✅ Ollama is running locally!")
    else:
        print("⚠️ Ollama is reachable but returned unexpected status.")
except requests.exceptions.ConnectionError:
    print("❌ Ollama is NOT running.")
    print("👉 Please open the Ollama app on your PC or run 'ollama serve' in a terminal.")
```

## 3. Pull Model (If needed)
Retrieving the `llama3.2:3b` model.

```python
# This runs the ollama command on your SYSTEM terminal
import subprocess

model_name = "llama3.2:3b"
print(f"⏳ Checking/Pulling model: {model_name}...")
subprocess.run(["ollama", "pull", model_name], check=True)
print(f"✅ Model {model_name} ready!")
```

## 4. Run Backend Server
Starting the FastAPI server on `localhost:8000`.

```python
import uvicorn
from backend import app

print("🚀 Starting Backend Server...")
print("👉 API URL: http://127.0.0.1:8000")

# Run server (blocking)
if __name__ == "__main__":
    uvicorn.run(app, host="127.0.0.1", port=8000)
```
